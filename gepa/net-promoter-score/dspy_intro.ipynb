### Try out DSPy on a simple example

```python

SMALL_MODEL_CANDIDATES: list[str] = [
    "gemini/gemini-2.5-flash-lite",
    "gemini/gemini-2.5-flash",
    "gemini/gemini-2.0-flash",
]

llm = dspy.LM(SMALL_MODEL_CANDIDATES[0])
dspy.configure(lm=llm)
dspy.configure_cache(enable_memory_cache=False, enable_disk_cache=False)
```

```python
predictor = dspy.Predict("question -> answer: int", cache = False)
predictor(
    question=(
        "I have 5 different balls and I randomly select 4. "
        "How many possible combinations of the balls I can get?"
    )
)
```

```python
# see the system and user messages and response in dspy field delimiter syntax
dspy.inspect_history(n = 1)  
```

```python
from dspy import PythonInterpreter
def evaluate_math(expr: str) -> str:
    # Executes Python and returns the output as string
    with PythonInterpreter() as interp:
        return interp(expr)

# dspy.ReAct is a multi-step, tool-calling module: 
# the agent loops thought → tool call → observation until it produces a final answer.
react_model = dspy.ReAct(
    signature="question -> answer: int",  # expects an int answer
    tools=[evaluate_math]
)
response = react_model(
    question=(
        "I have 25 different balls and I randomly select 9. "
        "How many possible combinations of the balls I can get?"
    )
)
print(response)
response.answer
```

```python
# response.trajectory is the flat dict that records every step of that loop, 
# with each key suffixed by the iteration index.
response.trajectory
```
